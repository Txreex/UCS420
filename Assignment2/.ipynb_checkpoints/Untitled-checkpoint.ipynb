```python
from typing import *
import random
from collections import Counter
import sys

def extract(rollno : int) -> list[int] :
    res : list[int] = []
    while(rollno != 0) :
        digit : int =  rollno%10
        rollno //= 10
        res.append(digit)
    res.reverse()
    return res

def highestScore(score : tuple[int , ...]) -> tuple[int , int] : 
    index : int = -1 
    maxScore : int  = max(score)
    
    for i in range(0 ,len(score)) : 
        if score[i] == maxScore : 
            index = i
            return (maxScore , index)
    return (-1 , -1)

def lowestScore(score : tuple[int , ...]) -> tuple[int , int] : 
    minScore : int = min(score)
    count : int = 0

    for i in range(0 , len(score)) : 
        if score[i] == minScore :
            count += 1;

    return (minScore , count);

def reverseTuple(nums : tuple) -> list :
    return list(nums[::-1])

def findFirstIndex(nums : tuple[int , ...] , score : int) -> int :
    for i in range(0 , len(nums)) : 
        if nums[i] == score :
            return i
    return -1

 
def isPrime(n : int) -> bool:
    if n < 2:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True
```

```python
def main(args : List[str]) -> None :
    ## 1
    print("QUESTION 1 : ")
    rollno : int = 1024170009
    L : List[int] = extract(rollno)
    print("L : " , L)
    
    L.append(70)      ## inserting 70 at the end of the list
    print("L.append(70) : " , L)
    L.insert(0 , -1)  ## inserting -1 at 0th index
    print("L.insert(0 , -1) : " , L)

    L.remove(70)      ## removing 70 from the list
    print("L.remove(70) : " , L)
    L.pop()           ## removing the last element
    print("L.pop() : " , L)

    L.sort()
    print("After sorting : " , L)
    L.sort(reverse = True)
    print("After sorting with reverse : " , L)

    print("First 3 elements :" , L[0:3])
    print("Last 3 elements  :" , L[-3:])

    avg : int = sum(L)/len(L)
    newL : List[int] = [x for x in L if x > avg]
    print("newList : " , newL)
    print("-------------------------------------------")


    ## 2
    print("QUESTION 2")
    scores : tuple[int] = tuple(L[0 : 8])
    print(scores)

    maxi : tuple[int] = highestScore(scores)
    print("Highest Score : " , maxi[0] , " , index : " , maxi[1])
    mini : tuple[int] = lowestScore(scores)
    print("Lowest Score : " , mini[0] , " , count : " , mini[1])

    scores_l : List[int] = reverseTuple(scores)
    print("Reversed tuple : " , scores_l)
    ## A tuple cannot be reversed because its an immutable datatype

    inputScore : int = int(input("Enter Score : "))
    index : int = findFirstIndex(scores , inputScore)
    if(index != -1) :
        print("Score Found at : " , index)
    else :
        print("Score Not Found")

    firstScore : int
    secondScore : int 
    remainingScores : list[int]

    firstScore , secondScore , *remainingScores = scores
    print("First Score : " , firstScore)
    print("Second Score : " , secondScore)
    print("Remaining Scores : " , remainingScores)
    print("-------------------------------------------")


    ## 3
    print("QUESTION 3")
    
    random.seed(rollno)
    
    randomNumbers : List[int] = [random.randint(100, 900) for _ in range(100)]
    
    print("Random Numbers : ", randomNumbers)
    
    # ii. Count odd numbers
    oddNumbers : List[int] = [x for x in randomNumbers if x % 2 != 0]
    print("Number of odd numbers : ", len(oddNumbers))
    
    # iii. Count even numbers
    evenNumbers : List[int] = [x for x in randomNumbers if x % 2 == 0]
    print("Number of even numbers : ", len(evenNumbers))
     
    
    primeNumbers : List[int] = [x for x in randomNumbers if isPrime(x)]
    
    print("Number of prime numbers : ", len(primeNumbers))
    print("Prime numbers : ", primeNumbers)
    
    
    # v. Most frequently occurring number
    frequency = Counter(randomNumbers)
    
    mostFrequentNumber : int
    mostFrequentCount : int
    
    mostFrequentNumber, mostFrequentCount = frequency.most_common(1)[0]
    
    print("Most frequent number : ", mostFrequentNumber)
    print("Frequency : ", mostFrequentCount)
    
    print("-------------------------------------------")

    digits : List[int] = extract(rollno)
    digits8 : List[int] = digits[:8]


    ## 4
    print("QUESTION 4")
    
    digits : List[int] = extract(rollno)
    digits8 : List[int] = digits[:8]
    
    A : set[int] = {digit * 7 for digit in digits8}
    B : set[int] = {digit * 9 for digit in digits8}
    
    print("Set A : ", A)
    print("Set B : ", B)
    
    # vi. Union
    unionSet = A.union(B)
    print("Union : ", unionSet)
    
    # vii. Intersection
    intersectionSet = A.intersection(B)
    print("Intersection : ", intersectionSet)
    
    # viii. Difference
    AminusB = A.difference(B)
    BminusA = B.difference(A)
    
    print("A - B : ", AminusB)
    print("B - A : ", BminusA)
    
    # difference() gives elements present in one set but not the other.
    # symmetric_difference() gives elements present in either set but not in both.
    
    # ix. Symmetric difference
    symmetricDifference = A.symmetric_difference(B)
    print("Symmetric Difference : ", symmetricDifference)
    
    # x. Subset and superset
    print("Is A subset of B : ", A.issubset(B))
    print("Is B superset of A : ", B.issuperset(A))
    
    # xi. discard()
    X : int = int(input("Enter value to remove from A : "))
    
    A.discard(X)
    
    print("Set A after discard : ", A)
    
    # discard() is safer than remove() because it does not raise an error
    # if the value is not present in the set.
    
    print("-------------------------------------------")


    ## 5
    print("QUESTION 5")
    
    my_dict = {
        "name": "Tanay Singh",
        "roll_no": "1024170009",
        "branch": "CSE",
        "age": 20,
        "city": "Amritsar"
    }
    
    print("Original Dictionary : ", my_dict)
    
    
    # i. Rename city to location
    my_dict["location"] = my_dict.pop("city")
    
    print("After renaming city to location : ", my_dict)
    

    # ii. Add CGPA
    my_dict["cgpa"] = 9.86
    
    print("After adding CGPA : ", my_dict)
    
    
    # iii. Increase age by 1
    my_dict["age"] += 1
    
    print("After increasing age : ", my_dict)
    
    
    # iv. Delete branch using pop()
    dict_pop = my_dict.copy()
    
    removedBranch = dict_pop.pop("branch")
    
    print("Dictionary after pop() : ", dict_pop)
    print("Value returned by pop() : ", removedBranch)
    
    
    # Delete branch using del
    dict_del = my_dict.copy()
    
    del dict_del["branch"]
    
    print("Dictionary after del : ", dict_del)
    
    # pop() removes the key-value pair and returns the value,
    # while del removes the key-value pair but does not return it.
    
    
    # v. Iterate using items()
    print("Key-value pairs : ")
    
    for key, value in my_dict.items():
        print(key, "→", value)
    
    
    # vi. Check whether email exists
    if "email" in my_dict:
        print("Email : ", my_dict["email"])
    else:
        print("Email key does not exist.")
    
    
    # vii. Friend dictionary
    friend_dict = {
        "name": "john",
        "roll_no": "1024170010",
        "branch": "CSE",
        "age": 20,
        "city": "Delhi"
    }
    
    merged_dict = {**my_dict, **friend_dict}
    
    print("Friend Dictionary : ", friend_dict)
    print("Merged Dictionary : ", merged_dict)
    
    # When both dictionaries have the same key,
    # the value from the second dictionary wins.
    
    
    # viii. Dictionary comprehension
    string_dict = {
        key: value
        for key, value in my_dict.items()
        if isinstance(value, str)
    }
    
    print("Dictionary containing only string values : ", string_dict)
    
    print("-------------------------------------------")
    
```

```python
if(__name__ == "__main__") :
    main(sys.argv)
```
