```python
data = {
    "Employee_ID":[101,102,103,104,105],
    "Name":["Alice","Bob","Charlie","Diana","Edward"],
    "Department":["HR","IT","IT","Marketing","Sales"],
    "Age":[29,34,41,28,38],
    "Salary":[50000,70000,65000,55000,60000],
    "Years_of_Experience":[4,8,10,3,12],
    "Joining_Date":["2020-03-15","2017-07-19","2013-06-01",
                    "2021-02-10","2010-11-25"],
    "Gender":["Female","Male","Male","Female","Male"],
    "Bonus":[5000,7000,6000,4500,5000],
    "Rating":[4.5,4.0,3.8,4.7,3.5]
}

df = pd.DataFrame(data)
print(df)
```

```python
print(df.shape)
print(df.info())
print(df.describe())
print(df.head())
print(df.tail(3))

avg_sal = df['Salary'].mean()
print(avg_sal)

total_bonus = df['Bonus'].sum()
print(total_bonus)

youngest_emp_age = df['Age'].min()
print(youngest_emp_age)

max_rating = df['Rating'].max()
print(max_rating)

df = df.sort_values(
    by = 'Salary',
    ascending = False
)
print(df)

def comp(rating : int) -> str: 
    if(rating >= 4.5) : return 'Excelent'
    if(rating >= 4.0) : return 'Good'
    return 'Average'

df["Performance"] = df["Rating"].apply(comp)
print(df)


print(df.isnull().sum())


df.rename(
    columns={"Employee_ID":"ID"},
    inplace=True
)

print(df[df["Years_of_Experience"] > 5])
print(df[df["Department"]=="IT"])

df["Tax"] = df["Salary"] * 0.10
print(df)

df.to_csv(
    "modified_employees.csv",
    index=False
)
```
