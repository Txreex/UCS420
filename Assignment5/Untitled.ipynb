```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Question 1
gfg = np.matrix('[4, 1, 9; 12, 3, 1; 4, 5, 6]')
print(gfg.sum())
print(gfg.sum(axis = 0)) # rows
print(gfg.sum(axis = 1))
```

```python
# Question 2(a)
array = np.array([10, 52, 62, 16, 16, 54, 453])

# i. Sorted array
sorted_array = np.sort(array)
print("Sorted array:", sorted_array)

# ii. Indices of sorted array
indices = np.argsort(array)
print("Indices of sorted array:", indices)

# iii. 4 smallest elements
smallest_4 = np.sort(array)[:4]
print("4 smallest elements:", smallest_4)

# iv. 5 largest elements
largest_5 = np.sort(array)[-5:]
print("5 largest elements:", largest_5)


# Question 2(b)
array = np.array([1.0, 1.2, 2.2, 2.0, 3.0, 2.0])

# Integer elements
integer_elements = array[array == array.astype(int)]
print("Integer elements:", integer_elements)

# Float elements only
float_elements = array[array != array.astype(int)]
print("Float elements:", float_elements)
```

```python
# Question 3(a)
X = ord('T') + ord('S')
sales = np.array([X, X+50, X+100, X+150, X+200])
print("X =", X)
print("Sales:", sales)

# Question 3(b)
tax_rate = ((X % 5) + 5) / 100
tax_amount = sales * tax_rate
sales_after_tax = sales + tax_amount
print("Tax rate:", tax_rate)
print("Tax amount:", tax_amount)
print("Sales after tax:", sales_after_tax)


# Question 3(c)
discount_rate = np.where(sales < X + 100, 0.05, 0.10)
discounted_sales = sales * (1 - discount_rate)
print("Discount rates:", discount_rate)
print("Sales after discount:", discounted_sales)

# Question 3(d)
weekly_sales = np.broadcast_to(sales, (3, 5))
print(weekly_sales)
week_increase = np.array([1.00, 1.02, 1.04]).reshape(3, 1)
adjusted_weekly_sales = weekly_sales * week_increase
print(adjusted_weekly_sales)
```

```python
# Question 4

# Generate x values
x = np.linspace(-10, 10, 100)

# Calculate y values
y1 = x ** 2
y2 = np.sin(x)
y3 = np.exp(x)
y4 = np.log(np.abs(x) + 1)

# Plot y = x^2
plt.figure(figsize=(8, 5))
plt.plot(x, y1)

plt.title("Plot of Y = x²")
plt.xlabel("x")
plt.ylabel("Y = x²")

plt.grid(True)
plt.show()
```

```python
plt.plot(x, y2)
plt.title("Plot of Y = sin(x)")
plt.xlabel("x")
plt.ylabel("Y = sin(x)")
plt.grid(True)
plt.show()
```

```python
plt.plot(x, y3)
plt.title("Plot of Y = e^x")
plt.xlabel("x")
plt.ylabel("Y = e^x")
plt.grid(True)
plt.show()
```

```python
plt.plot(x, y4)
plt.title("Plot of Y = log(|x| + 1)")
plt.xlabel("x")
plt.ylabel("Y")
plt.grid(True)
plt.show()
```
