```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

corrected_temperature = temperature + 2
print("Corrected temperatures:", corrected_temperature)

fahrenheit = (9 / 5) * temperature + 32
print("Temperature in Fahrenheit:", fahrenheit)

greater_than_32 = temperature[temperature > 32]
print("Readings greater than 32°C:", greater_than_32)

count = np.sum(temperature > 32)
print("Count:", count)


steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print("\nTotal steps:", np.sum(steps))

print("Mean steps:", np.mean(steps))


print("Maximum:", np.max(steps))
print("Minimum:", np.min(steps))

# d. Total steps for each day
print("Daily totals:", np.sum(steps, axis=0))

# e. Total steps for each user
print("User totals:", np.sum(steps, axis=1))

# f. Position of maximum value
position = np.unravel_index(np.argmax(steps), steps.shape)
print("Position of maximum:", position)


# a. Original array
original = np.array([1, 2, 3, 4, 5, 6])

# b. Slice from index 1 to 4
subset = original[1:5]
print("\nSubset:", subset)

# c. Modify first element of subset
subset[0] = 999

print("Original after modifying subset:", original)
print("Subset:", subset)

# d. Slice using copy()
copied_array = original[1:5].copy()
copied_array[0] = 500

print("Original after modifying copy:", original)
print("Copied array:", copied_array)

# e. Create 1 to 12 and reshape into 3x4
matrix = np.arange(1, 13).reshape(3, 4)

print("\nMatrix:")
print(matrix)

# f. Indexing and slicing
print("First row:", matrix[0])
print("Last row:", matrix[-1])
print("Second column:", matrix[:, 1])
print("Rows 1-2 and columns 2-3:")
print(matrix[0:2, 1:3])

# g. flatten() and ravel()
flattened = matrix.flatten()
raveled = matrix.ravel()

print("Flatten:", flattened)
print("Ravel:", raveled)

# h. Modify ravel()
raveled[0] = 999

print("\nAfter modifying ravel:")
print("Matrix:")
print(matrix)

# i. Modify flatten()
flattened = matrix.flatten()
flattened[1] = 500

print("\nAfter modifying flatten:")
print("Flattened:", flattened)
print("Matrix:")
print(matrix)

# j. Shape, dimensions, size and dtype
print("\nShape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)



X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

# a. Shape and dimensions
print("\nShape of X:", X.shape)
print("Dimensions of X:", X.ndim)

# b. Transpose
print("\nX Transpose:")
print(X.T)

# c. X.T @ X
XtX = X.T @ X
print("\nX.T @ X:")
print(XtX)

# d. Matrix inverse
inverse = np.linalg.inv(XtX)
print("\nInverse of X.T @ X:")
print(inverse)

# e. Ordinary Least Squares
beta = np.linalg.inv(X.T @ X) @ X.T @ y
print("\nOLS Coefficients:")
print(beta)

# g. Predict assistance score for new user
new_user = np.array([5, 40, 7])

predicted_score = new_user @ beta

print("\nPredicted Assistance Score:", predicted_score)
```
